```python
import os
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image
```

```python
# 1. Define Paths
BASE_DIR = r"D:\Traffic Sign Classification using CNN\Datasets\GTSRB"
TRAIN_CSV = os.path.join(BASE_DIR, "Train.csv")

# 2. Load the metadata CSV
df_train = pd.read_csv(TRAIN_CSV)
print("--- Dataset Sample ---")
print(df_train.head())

# 3. Check total images and unique traffic sign classes
print(f"\nTotal Training Images: {len(df_train)}")
print(f"Total Unique Classes: {df_train['ClassId'].nunique()}")

```

```python
# 4. Plot Class Distribution (To see if the dataset is imbalanced)
plt.figure(figsize=(12, 5))
df_train['ClassId'].value_counts().sort_index().plot(kind='bar', color='skyblue')
plt.title("GTSRB Class Distribution")
plt.xlabel("Class ID")
plt.ylabel("Number of Images")
plt.show()
```

```python
# 5. Visualize a few random samples
fig, axes = plt.subplots(1, 5, figsize=(15, 3))
samples = df_train.sample(5)

for i, (_, df_row) in enumerate(samples.iterrows()):
    # Reconstruct the absolute path to the image
    img_path = os.path.join(BASE_DIR, df_row['Path'])
    img = Image.open(img_path)
    
    axes[i].imshow(img)
    axes[i].set_title(f"Class: {df_row['ClassId']}")
    axes[i].axis('off')

plt.tight_layout()
plt.show()
```
